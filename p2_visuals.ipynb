This file contains the EDA of the cleaned dataset. Here we will analyze the dataset to uncover the trends and patterns.

Step 1: Import the file and understand the basics of the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_sales_data.csv")
```

```python
print(df.head())
print(df.shape)
print(df.columns)
print(df.describe())
```

Step 2: Create visuals to analyze the basic trends.
First visual:Sales Distribution

```python
plt.hist(df['Sales'], bins=30)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

Second Visual:Category Count

```python
df['Product_Category'].value_counts().plot(kind='bar')
plt.title("Product Category Distribution")
plt.show()
```

Visual 3: Sales by region

```python
df.groupby('Region')['Sales'].sum().plot(kind='bar')
plt.title("Sales by Region")
plt.show()
```

Visual 4: Profit by category.

```python
df.groupby('Product_Category')['Profit'].sum().plot(kind='bar')
plt.title("Profit by Category")
plt.show()
```

Visual 5: Sales versus profit

```python
plt.scatter(df['Sales'], df['Profit'])
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

Visual 6: Correlation between sales, profit, quantity, discount.

```python
correlation = df[['Sales', 'Profit', 'Quantity', 'Discount']].corr()
print(correlation)
```

Top ten customers

```python
top_customers = df.groupby('Customer_Name')['Sales'].sum().sort_values(ascending=False).head(10)
print(top_customers)
```

Top ten products.

```python
top_products = df.groupby('Product_Name')['Sales'].sum().sort_values(ascending=False).head(10)
print(top_products)
```

All these visuals and tables show trends and behaviour of the dataset and can be used to draw insights into the data.

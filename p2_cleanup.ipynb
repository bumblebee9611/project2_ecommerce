This file shows cleanup of data in steps for further analysis.
Step 1: importing the file and checking if import is successfull.

```python
import pandas as pd

df = pd.read_csv("ecommerce_sales_dataset.csv")
df.head()
```

Step 2: Check for duplicates and null values.

```python
print("Null summary")
print(df.isnull().sum())
print("Duplicate summary")
print(df.duplicated().sum())
```

We can clearly see that null values as well as duplicate values are present.

Step 3: We will remove duplicate rows and check if they have been removed.

```python
print("before removing duplicates")
print(df.duplicated().sum())
df = df.drop_duplicates()
print("after removing duplicates")
print(df.duplicated().sum())
```

Step 4: Now we will look for null values and fill them with median value for numeric type and unknown for text type.

```python
print("Null summary before")
print(df.isnull().sum())

df['Sales'] = df['Sales'].fillna(df['Sales'].median())
df['Profit'] = df['Profit'].fillna(df['Profit'].median())
df['Customer_Name'] = df['Customer_Name'].fillna("Unknown")

print("Null summary after")
print(df.isnull().sum())
```

We are done with cleanup, now we just need to save the cleaned dataset.

```python
df.to_csv("cleaned_sales_data.csv", index=False)
```
